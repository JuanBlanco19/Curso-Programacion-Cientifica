## Ejercicio 1 — Ficha de la campaña de medición

tipo de dato exacto requerido ('str', 'int', 'float', 'bool',). finalmente,

## Ejercicio 2 — Conversión de unidades y error frente al equipo patrón

Este ejercicio convierte la lectura del sensor de mg/m³ a µg/m³ usando el factor de conversion previamente definido. Ademas, calcula las metricas de
error (absoluto, relativo porcentual y cuadratico) respecto al valor medido por el equipo patron del laboratorio.

```python
#Declaracion de variables iniciales

lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

#Conversion de unidades y calculo de errores

lectura_sensor_ug = lectura_sensor_mg * factor_coversion
error_absoluto = valor_patron_ug - lectura_sensor_ug
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100
error_cuadratico = error_absoluto ** 2

#Impresion de resultados

print("Lectura del sensor en µg/m³:", lectura_sensor_ug)
print("Error absoluto:", error_absoluto)
print("Error relativo porcentual (%):", error_relativo_porcentual)
print("Error cuadrático:", error_cuadratico)
```

## Ejercicio 3 — Cobertura temporal del lote descargado

En este ejercicio se calcula la cobertura temporal total del lote de mediciones descargadas utilizando unicamente operadores enteros ('//' y '%'). se obtiene el equivalente del numero de lecturas en dias, horas y lecturas sobrantes.

```python
#Frecuencia de lecturas del sensor

lecturas_por_horas = 12

#Calculos de tiempo con divison entera y modulo

horas_completas = numero_lecturas // lecturas_por_horas
lecturas_sobrantes = numero_lecturas % lecturas_por_horas

dias_completas = horas_completas // 24
horas_sobrantes = horas_completas % 24

#Impresion de resultados

print("Horas completas:", horas_completas)
print("Lecturas sobrantes:", lecturas_sobrantes)
print("Días completos:", dias_completas)
print("Horas sobrantes:", horas_sobrantes)
```

## Ejercicio 4 — Validar una lectura sin decidir todavía

Este ejercicio evalua la validez de una lectura comprobando mediante operadores logicos ('and', 'not') y de comparacion que el dato no correponda al codigo centinela de falta ('-999.0') y que se encuntre dentro del rango fisico aceptable del sensor [0, 500] se realiza sin utilizar estructuras de control 'if'.

```python
#Variables de la lectura y umbrales
lectura = 41.8
codigo_dato_faltante = -999.0
limite_superior_sensor = 500.0

#Evaluacion logica sin usar if

es_dato_faltante = lectura == codigo_dato_faltante
esta_en_rango = (lectura >= 0) and (lectura <= limite_superior_sensor)
lectura_valida = esta_en_rango and (not es_dato_faltante)

#Impresion de resultados

print("es dato faltante:", es_dato_faltante)
print("esta en rango:", esta_en_rango)
print("lectura valida:", lectura_valida)
```

## Ejercicio 6 — Limpiar la serie horaria
Este ejercicio recorre una serie de 12 lecturas consecutivas para filtrar las mediciones inválidas (centinela `-999.0`) haciendo uso de `continue`. Acumula las lecturas válidas en una nueva lista, contabiliza los descartes y calcula el promedio final de PM2.5 sobre los datos limpios.

```python
# Datos de entrada
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]

# Inicialización de estructuras y acumuladores
lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

# Filtrado y acumulación
for lectura in lecturas_pm25:
    if lectura == -999.0:
        lecturas_descartadas += 1
        continue

    lecturas_validas.append(lectura)
    suma_pm25 += lectura

# Cálculo del promedio posterior al bucle
promedio_pm25 = suma_pm25 / len(lecturas_validas)

# Mostrar resultados
print("Lecturas válidas:", len(lecturas_validas))
print("Lecturas descartadas:", lecturas_descartadas)
print("Promedio PM2.5:", promedio_pm25)
```

## Ejercicio 7 — Estadística descriptiva de la serie limpia
En este ejercicio se efectúa la descripción estadística de las lecturas válidas utilizando dos recorridos `for` independientes sin recurrir a funciones integradas como `max()`, `min()` o `sum()`. El primer recorrido determina el máximo, mínimo y rango; el segundo calcula la desviación estándar sobre el promedio obtenido en el ejercicio anterior.

```python
# 1. Primer recorrido: Obtención de máximo y mínimo
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for lectura in lecturas_validas:
    if lectura > maximo_pm25:
        maximo_pm25 = lectura
    if lectura < minimo_pm25:
        minimo_pm25 = lectura

rango = maximo_pm25 - minimo_pm25

# 2. Segundo recorrido: Cálculo de la dispersión
suma_desviaciones = 0.0
for lectura in lecturas_validas:
    suma_desviaciones += (lectura - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

# Mostrar resultados
print("Máximo PM2.5:", maximo_pm25)
print("Mínimo PM2.5:", minimo_pm25)
print("Rango:", rango)
print("Desviación estándar:", desviacion_estandar)
```

## Ejercicio 8 — Decaimiento tras cerrar la fuente de emisión
Este ejercicio simula el decaimiento progresivo del pico de contaminación (58.3 µg/m³) a una tasa de reducción del 15 % por hora (factor 0.85) utilizando un bucle `while`. Se calcula el número de horas necesarias para retornar a la categoría "Buena" (≤ 12.0 µg/m³).

```python
# Inicialización de variables
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0
horas_transcurridas = 0

# Simulación de decaimiento con while
while concentracion > umbral_buena:
    concentracion *= factor_decaimiento
    horas_transcurridas += 1

# Mostrar resultados
print("Horas transcurridas:", horas_transcurridas)
print("Concentración final:", concentracion)
```

## Ejercicio 9 — Registros de varias estaciones
En este ejercicio se procesa una lista de diccionarios que representa múltiples estaciones de la red de monitoreo. Para cada estación se desempaqueta la tupla de coordenadas, se calcula el promedio de sus lecturas mediante un bucle interno, se extrae la humedad de forma segura con `.get()` y se clasifica su calidad del aire en la escala oficial.

```python
# Definición de la estructura de datos
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte",  "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur",    "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

# Procesamiento de cada estación
for estacion in registros_estaciones:
    # Desempaquetado de coordenadas
    latitud, longitud = estacion["coordenadas"]

    # Cálculo del promedio de lecturas con for interno
    suma_estacion = 0.0
    for lect in estacion["lecturas"]:
        suma_estacion += lect
    promedio_estacion = suma_estacion / len(estacion["lecturas"])

    # Lectura de humedad segura con get
    humedad = estacion.get("humedad_relativa", "no disponible")

    # Clasificación del promedio
    if promedio_estacion <= 12.0:
        categoria_est = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_est = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_est = "Dañina para grupos sensibles"
    else:
        categoria_est = "Dañina para la salud"

    # Salida por pantalla por estación
    print(f"Código: {estacion['codigo']}, Sector: {estacion['sector']},Latitud: {latitud}, Longitud: {longitud}, Promedio: {promedio_estacion:.2f}, Categoría: {categoria_est}, Humedad: {humedad}")
```
