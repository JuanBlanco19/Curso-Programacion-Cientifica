#EJERCICIOS
##Ejercicio 1
Inventario de la tienda

```python
# 1. Declaración de las 7 variables con sus tipos de datos correspondientes
nombre_producto = "Cuaderno"  # str  (nombre de un producto a elección)
precio_cuaderno = 3500         # int  (precio entero)
precio_lapiz = 800             # int  (precio entero)
precio_cafe = 6200.50          # float (precio con decimales)
hay_descuento = False          # bool (indicador booleano)
cantidad_cuadernos = 3         # int  (cantidad entera)
cantidad_lapices = 5           # int  (cantidad entera)

# 2. Verificación del tipo de dato de cada variable con type()
print("nombre_producto:", type(nombre_producto))
print("precio_cuaderno:", type(precio_cuaderno))
print("precio_lapiz:", type(precio_lapiz))
print("precio_cafe:", type(precio_cafe))
print("hay_descuento:", type(hay_descuento))
print("cantidad_cuadernos:", type(cantidad_cuadernos))
print("cantidad_lapices:", type(cantidad_lapices))
```

##Ejercicio 2
Total de la compra y descuento

```python
# 1. Cálculo de subtotales para cuadernos y lápices
total_cuadernos = precio_cuaderno * cantidad_cuadernos
print("Total cuadernos:", total_cuadernos)

total_lapices = precio_lapiz * cantidad_lapices
print("Total lápices:", total_lapices)

# 2. Cálculo del total general (sumando cuadernos, lápices y el café)
total_general = total_cuadernos + total_lapices + precio_cafe
print("Total general:", total_general)

# 3. Cálculo del descuento del 10% y el valor final
descuento = total_general * 0.10
total_con_descuento = total_general - descuento

# 4. Impresión de los resultados finales
print("Descuento (10%):", descuento)
print("Total con descuento:", total_con_descuento)
```

##Ejercicio 3
Reparto en paquetes

```python
# 1. Cantidad de paquetes completos de 4 cuadernos (división entera)
paquetes_completos = cantidad_cuadernos // 4

# 2. Cuadernos sueltos restantes (módulo)
cuadernos_sueltos = cantidad_cuadernos % 4

# 3. Precio del café elevado al cuadrado (potencia)
precio_cafe_cuadrado = precio_cafe ** 2

# 4. Impresión de los tres resultados
print("Paquetes completos de 4 cuadernos:", paquetes_completos)
print("Cuadernos sueltos restantes:", cuadernos_sueltos)
print("Precio del café al cuadrado:", precio_cafe_cuadrado)
```

##Ejercicio 5
Promedio y comparacion de temperatura

```python
# 1. declaracion de variables de temperatura ( float)
temp_lunes = 18.5
temp_martes = 21.0
temp_miercoles = 19.8
temp_promedio_semana = 20.0

# 2. promedio de los tres dias
promedio_dias = (temp_lunes + temp_martes + temp_miercoles) / 3
print("promedio de los tres dias:", promedio_dias)

# 3. diferencia entre la temperatura del martes y la del lunes
diferencia_martes_lunes = temp_martes - temp_lunes
print("diferencia entre martes y lunes:", diferencia_martes_lunes)

# 4. evaluacion: el martes fue mas caluroso que el promedio de referencia
martes_mas_caluroso = temp_martes > temp_promedio_semana
print("el martes fue mas caluroso que el promedio:", martes_mas_caluroso)

# 5. evaluacion: el miercoles fue igual al promedio de referencia
miercoles_igual_promedio = temp_miercoles == temp_promedio_semana
print("el miercoles fue igual al promedio:", miercoles_igual_promedio)

# 6. evalucion: el lunes fue el dia mas frio de los tres
lunes_mas_frio = (temp_lunes < temp_martes) and (temp_lunes < temp_miercoles)
print("el lunes fue el dia mas frio de los tres:", lunes_mas_frio)
```

##Ejercicio 6
Temperatura templada y dia caluroso

```python
temp_lunes = 18.5
temp_martes = 21.0
temp_miercoles = 19.8

# 1. evaluacion con 'and' el lunes tuvo temperatura templada (entre 15 y 25 °C)
lunes_templado = (temp_lunes >= 15) and (temp_lunes <= 25)
print("el lunes tuvo temperatura templada:", lunes_templado)

# 2. evaluacion con 'or' al menos uno de los tres dias los 20 °C
al_menos_un_dia_caluroso = (temp_lunes > 20) or (temp_martes > 20) or (temp_miercoles > 20)
print("al menos uno de los tres dias supero los 20 °C:", al_menos_un_dia_caluroso)
```

##Ejercicio 10
Intercambiar dos precios

```python
# 1. Valores iniciales con los precios erróneos
precio_tornillo = 250
precio_tuerca = 180

# 2. Impresión de valores antes del intercambio
print("--- Antes del intercambio ---")
print("Precio tornillo:", precio_tornillo)
print("Precio tuerca:", precio_tuerca)

# 3. Intercambio utilizando la variable temporal
temporal = precio_tornillo
precio_tornillo = precio_tuerca
precio_tuerca = temporal

# 4. Impresión de valores después del intercambio
print("\n--- Después del intercambio ---")
print("Precio tornillo:", precio_tornillo)
print("Precio tuerca:", precio_tuerca)
```
