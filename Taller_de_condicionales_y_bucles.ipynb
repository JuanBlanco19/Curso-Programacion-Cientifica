## Ejercicio 1 — Clasificar el stock de un producto

```python
# 1. Declarar la variable con el valor indicado
cantidad_en_stock = 3

# 2. Clasificar el stock usando la estructura if / elif / else
if cantidad_en_stock == 0:
    nivel = "Agotado"
elif cantidad_en_stock < 5:
    nivel = "Stock bajo"
else:
    nivel = "Stock suficiente"

# 3. Mostrar el resultado
print(nivel)
```

## Ejercicio 2 — Alerta de reposición inmediata

```python
# Retomamos las variables del Ejercicio 1 (cantidad_en_stock y nivel)
# Evaluar con un operador lógico 'or' dentro de la condición del if
if nivel == "Agotado" or cantidad_en_stock < 5:
    print("Reponer pronto")
else:
    print("Sin acción requerida")
```

## Ejercicio 3 — Clasificar el clima del lunes

## Ejercicio 5 — Aprobación con mensaje

```python
# 1. Declarar las tres notas parciales con los valores indicados
nota_1 = 4.2
nota_2 = 3.5
nota_3 = 4.8

# 2. Calcular el promedio de las tres notas
promedio = (nota_1 + nota_2 + nota_3) / 3

# 3. Clasificar el resultado usando if / elif / else
if promedio < 3.0:
    resultado = "Reprobado"
elif promedio <= 4.4:  # Cubre el rango de 3.0 a 4.4 (inclusive)
    resultado = "Aprobado"
else:  # Para 4.5 o más
    resultado = "Aprobado con excelencia"

# 4. Mostrar el resultado
print(resultado)
```

## Ejercicio 6 — Recorrer el inventario completo

```python
# 1. Declarar la lista de cantidades en stock
cantidades_en_stock = [0, 3, 12, 0, 7, 2]

# 2. Recorrer la lista por índice usando for con range(len(...))
for indice in range(len(cantidades_en_stock)):
    # Obtener la cantidad de la posición actual
    cantidad = cantidades_en_stock[indice]

    # Clasificar la cantidad según la lógica del Ejercicio 1
    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    # Mostrar el mensaje con el índice y el nivel obtenido
    print(f"Producto {indice}: {nivel}")
```

## Ejercicio 7 — Promedio semanal de temperaturas

```python
# 1. Declarar la lista de temperaturas de la semana
temperaturas_semana = [18.5, 21.0, 19.8, 24.3, 17.2]

# 2. Declarar acumulador y contador inicializados en 0
suma_temperaturas = 0
dias_calurosos = 0

# 3. Recorrer la lista usando for con range(len(...))
for i in range(len(temperaturas_semana)):
    # Sumar cada temperatura al acumulador
    suma_temperaturas += temperaturas_semana[i]

    # Incrementar el contador si la temperatura es mayor a 20
    if temperaturas_semana[i] > 20:
        dias_calurosos += 1

# 4. Calular el promedio dividiendo la suma entre el total de días
promedio_semanal = suma_temperaturas / len(temperaturas_semana)

# 5. Mostrar los resultados
print(f"Promedio semanal: {promedio_semanal:.2f}°C")
print(f"Días calurosos: {dias_calurosos}")
```

## Ejercicio 8 — Comprar tornillos mientras alcance el presupuesto

```python
# 1. Declarar las variables con los valores indicados
presupuesto_ferreteria = 3000
precio_tornillo = 250

# 2. Declarar el contador inicializado en 0
tornillos_comprados = 0

# 3. Usar while para repetir mientras alcance el dinero
while presupuesto_ferreteria >= precio_tornillo:
    presupuesto_ferreteria -= precio_tornillo  # Descontar el precio del presupuesto
    tornillos_comprados += 1                  # Incrementar la cantidad comprada

# 4. Mostrar los resultados
print(f"Tornillos comprados: {tornillos_comprados}")
print(f"Presupuesto restante: {presupuesto_ferreteria}")
```

## Ejercicio 9 — Buscar la primera herramienta cara

```python
# 1. Declarar la lista con los precios de las herramientas
precios_herramientas = [12000.0, 8500.0, 25000.0, 9800.0]

# 2. Recorrer la lista directamente por sus valores (sin range ni índices)
for precio in precios_herramientas:
    # Evaluar si el precio supera los 20000
    if precio > 20000:
        print(f"¡Primera herramienta cara encontrada! Precio: {precio}")
        break  # Detiene el bucle de inmediato
    else:
        print(f"Sigue revisando... Precio actual: {precio}")
```

## Ejercicio 10 — Revisión completa del inventario de la ferretería

```python
# 1. Declarar las dos listas paralelas con los datos
nombres_herramientas = ["Martillo", "Taladro", "Sierra", "Destornillador", "Nivel", "Llave inglesa"]
cantidades_herramientas = [5, 0, 2, 8, 0, 1]

# 2. Recorrer ambas listas por índice con un bucle for y range(len(...))
for i in range(len(nombres_herramientas)):
    nombre = nombres_herramientas[i]
    cantidad = cantidades_herramientas[i]

    # 3. Clasificar la cantidad en la variable 'nivel'
    if cantidad == 0:
        nivel = "Agotado"
    elif cantidad < 5:
        nivel = "Stock bajo"
    else:
        nivel = "Stock suficiente"

    # 4. Aplicar la lógica del ejercicio según el nivel obtenido:

    # - Si es "Stock suficiente", no requiere atención: usar continue para saltar
    if nivel == "Stock suficiente":
        continue

    # - Si es "Agotado", imprimir alerta urgente y detener el bucle con break
    elif nivel == "Agotado":
        print(f"ALERTA REPOSICIÓN URGENTE: {nombre} está agotado")
        break

    # - Si es "Stock bajo", imprimir la advertencia y continuar con normalidad
    elif nivel == "Stock bajo":
        print(f"Stock bajo en {nombre}: quedan {cantidad} unidades")
```
