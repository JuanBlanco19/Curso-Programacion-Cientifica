### Ejercicio 1 — Calibrar una lectura
Se define la función `calibrar_lectura` con valores por defecto en su firma (`pendiente=1.0`, `intercepto=0.0`)[cite: 2]. Utiliza `return` para entregar el resultado de la recta de calibración y se realizan las llamadas solicitadas empleando argumentos por posición y por nombre.

```python
def calibrar_lectura(valor: float, pendiente: float = 1.0, intercepto: float = 0.0) -> float:
    """Aplica la recta de calibración a una lectura dada."""
    return valor * pendiente + intercepto

sin_calibrar = calibrar_lectura(10.0)
calibrada = calibrar_lectura(10.0, 1.2, -0.5)
con_intercepto = calibrar_lectura(10.0, intercepto=2.0)

print("sin_calibrar:", sin_calibrar)
print("calibrada:", calibrada)
print("con_intercepto:", con_intercepto)
```

### Ejercicio 2 — ¿Es utilizable la lectura?
Se declara la lista oficial de lecturas crudas `lecturas_sbc_a` con los valores originales de la guía[cite: 2]. Se implementa la función pura `es_lectura_valida` utilizando comparación encadenada y operadores lógicos para validar que el dato no sea el centinela `-999.0` y se encuentre en el rango físico de `0` a `500.0` inclusive[cite: 2].

```python
lecturas_sbc_a = [10.0, -999.0, 25.0, 40.0, -999.0, 8.0, 52.0, 600.0]

def es_lectura_valida(valor: float) -> bool:
    """Verifica si una lectura no es centinela y se encuentra en el rango físico [0, 500]."""
    if valor == -999.0:
        return False
    return 0.0 <= valor <= 500.0

print("25.0 es válida:", es_lectura_valida(25.0))
print("-999.0 es válida:", es_lectura_valida(-999.0))
print("600.0 es válida:", es_lectura_valida(600.0))
print("0.0 es válida:", es_lectura_valida(0.0))
```

### Ejercicio 3 — Promedio de las lecturas útiles
Se define la función `promedio_valido` que reutiliza estrictamente `es_lectura_valida` mediante un bucle `for` con variables locales, calculando el promedio de forma dinámica[cite: 2]. Si no hay lecturas válidas, retorna `None` de forma segura[cite: 2].

```python
def promedio_valido(lecturas: list[float]) -> float | None:
    """Calcula el promedio dinámico de las lecturas que resulten válidas."""
    acumulador = 0.0
    contador = 0
    for l in lecturas:
        if es_lectura_valida(l):
            acumulador += l
            contador += 1
    if contador == 0:
        return None
    return acumulador / contador

promedio_sbc_a = promedio_valido(lecturas_sbc_a)

print("promedio_sbc_a:", promedio_sbc_a)
print("Promedio sin lecturas válidas:", promedio_valido([-999.0, -999.0]))
```

### Ejercicio 4 — Calibrar la serie con map, filter y lambda
Se procesa la serie de datos aplicando exclusivamente programación funcional (`filter`, `map` y funciones `lambda`), sin emplear bucles ni comprensiones de lista[cite: 2]. Se filtran las lecturas útiles, se calibran con los parámetros de SBC-A y se calculan dinámicamente las excedencias sobre `35.4`[cite: 2].

```python
pendiente_sbc_a = 1.2
intercepto_sbc_a = -0.5

lecturas_utiles = filter(es_lectura_valida, lecturas_sbc_a)
lecturas_calibradas = list(map(lambda x: calibrar_lectura(x, pendiente_sbc_a, intercepto_sbc_a), lecturas_utiles))
excedencias = len(list(filter(lambda x: x > 35.4, lecturas_calibradas)))

print("lecturas_calibradas:", lecturas_calibradas)
print("excedencias:", excedencias)
```

### Ejercicio 5 — La clase Lectura
Se define la clase `Lectura` con un método constructor `__init__` que incorpora los valores por defecto requeridos (unidad `"ug/m3"` y código de sensor `"sin asignar"`)[cite: 2]. Se implementan los métodos de instancia `a_microgramos` y `es_dato_faltante`, instanciando los objetos de prueba pedidos por la guía[cite: 2].

```python
class Lectura:
    """Representa una medición individual de contaminantes atmosféricos."""

    def __init__(self, valor: float, unidad: str = "ug/m3", codigo_sensor: str = "sin asignar"):
        """Inicializa una lectura guardando su valor, unidad y código de sensor."""
        self.valor = float(valor)
        self.unidad = unidad
        self.codigo_sensor = codigo_sensor

    def a_microgramos(self) -> float:
        """Convierte y retorna el valor de la lectura en microgramos por metro cúbico."""
        if self.unidad == "mg/m3":
            return self.valor * 1000.0
        return self.valor

    def es_dato_faltante(self) -> bool:
        """Indica si el valor corresponde al código centinela de fallo (-999.0)."""
        return self.valor == -999.0

lectura_campo = Lectura(0.0252, "mg/m3", "SBC-A")
lectura_faltante = Lectura(-999.0, codigo_sensor="SBC-A")
lectura_normal = Lectura(30.0)

print("lectura_campo en ug/m3:", lectura_campo.a_microgramos())
print("lectura_faltante es dato faltante:", lectura_faltante.es_dato_faltante())
print("lectura_campo es dato faltante:", lectura_campo.es_dato_faltante())
print("unidad y código de lectura_normal:", lectura_normal.unidad, lectura_normal.codigo_sensor)
```

### Ejercicio 6 — La clase Sensor: validación y encapsulamiento
Se construye la clase `Sensor` aplicando validaciones estrictas con `raise ValueError` en el constructor antes de asignar cualquier atributo si el código está vacío o la pendiente es $\le 0$[cite: 2]. Los parámetros de calibración se encapsulan de forma interna como `_pendiente` e `_intercepto`, e implementa los métodos mágicos `__repr__` y `__str__` junto con la reutilización de `calibrar_lectura`[cite: 2].

```python
class Sensor:
    """Representa un sensor de calibración con atributos protegidos y validación."""

    def __init__(self, codigo: str, pendiente: float = 1.0, intercepto: float = 0.0):
        """Inicializa el sensor validando parámetros antes de almacenarlos."""
        if not codigo or not codigo.strip():
            raise ValueError("El código del sensor no puede ser un texto vacío.")
        if pendiente <= 0:
            raise ValueError("La pendiente de calibración debe ser estrictamente mayor a cero.")

        self.codigo = codigo
        self._pendiente = float(pendiente)
        self._intercepto = float(intercepto)

    def calibrar(self, valor: float) -> float:
        """Calibra un valor empleando la función externa calibrar_lectura."""
        return calibrar_lectura(valor, self._pendiente, self._intercepto)

    def __repr__(self) -> str:
        """Devuelve la representación formal exacta del sensor."""
        return f"Sensor(codigo='{self.codigo}', pendiente={self._pendiente}, intercepto={self._intercepto})"

    def __str__(self) -> str:
        """Devuelve una descripción legible en una sola línea del sensor."""
        return f"Sensor {self.codigo} (pendiente: {self._pendiente}, intercepto: {self._intercepto})"

sensor_a = Sensor("SBC-A", 1.2, -0.5)
sensor_b = Sensor("SBC-B")

print("sensor_a repr:", repr(sensor_a))
print("print(sensor_a):", sensor_a)
print("Lista de sensores:", [sensor_a, sensor_b])
print("Calibrar 10.0 con sensor_a:", sensor_a.calibrar(10.0))
print("Calibrar 10.0 con sensor_b:", sensor_b.calibrar(10.0))
```

##Ejercicio 7 — La clase Estacion: objetos que contienen objetos
La estacion EST-04 guarda sus lecturas como objetos Lectura (Ej. 5) en _lecturas.
valores() obtiene cada numero llamando a a_microgramos() de cada lectura (no toca
valor ni unidad directamente). agregar() anade una lectura; fuera de la clase solose usan metodos, nunca _lecturas. Los valores coinciden con lecturas_sbc_a del Ej. 2.

```python
class Estacion:
    """Agrupa colecciones de objetos Lectura asociados a una estación y sector."""

    def __init__(self, codigo: str, sector: str):
        """Inicializa la estación con su código, sector y una lista interna vacía."""
        self.codigo = codigo
        self.sector = sector
        self._lecturas: list[Lectura] = []

    def agregar(self, lectura: Lectura) -> None:
        """Agrega un objeto Lectura a la estación."""
        self._lecturas.append(lectura)

    def valores(self) -> list[float]:
        """Extrae la concentración en ug/m3 de cada lectura usando a_microgramos()."""
        return [l.a_microgramos() for l in self._lecturas]

    def __repr__(self) -> str:
        """Devuelve la representación formal requerida para la estación."""
        return f"Estacion('{self.codigo}', '{self.sector}', lecturas={len(self._lecturas)})"

datos_campana = [
    (10.0, "ug/m3"), (-999.0, "ug/m3"), (0.025, "mg/m3"), (40.0, "ug/m3"),
    (-999.0, "ug/m3"), (8.0, "ug/m3"), (52.0, "ug/m3"), (600.0, "ug/m3"),
]

estacion_oriente = Estacion("EST-04", "Oriente")

for valor, unidad in datos_campana:
    lect_obj = Lectura(valor, unidad=unidad, codigo_sensor="SBC-A")
    estacion_oriente.agregar(lect_obj)

print("estacion_oriente repr:", repr(estacion_oriente))
print("Valores de la estación:", estacion_oriente.valores())
```

### Ejercicio 8 — Integración: el informe de la estación
Se define la función oficial `categoria_pm25` basada en una única estructura `if / elif / else` estricta[cite: 2]. Se crea la clase `InformeEstacion` cuyos métodos integran dinámicamente las funciones auxiliares de validación, calibración y promediado, implementando el método `__str__` para formatear exactamente el reporte consolidado de la estación[cite: 2].

```python
def categoria_pm25(concentracion: float) -> str:
    """Clasifica la concentración de PM2.5 según la escala oficial del taller."""
    if concentracion <= 12.0:
        return "Buena"
    elif concentracion <= 35.4:
        return "Moderada"
    elif concentracion <= 55.4:
        return "Dañina para grupos sensibles"
    else:
        return "Dañina para la salud"

class InformeEstacion:
    """Genera un informe analítico integrado combinando estación y sensor."""

    def __init__(self, estacion: Estacion, sensor: Sensor):
        """Inicializa el informe guardando la estación y el sensor en atributos internos."""
        self._estacion = estacion
        self._sensor = sensor

    def promedio_calibrado(self) -> float | None:
        """Calcula el promedio calibrado filtrando y transformando las lecturas válidas de la estación."""
        vals = self._estacion.valores()
        filtradas = filter(es_lectura_valida, vals)
        calibradas = list(map(self._sensor.calibrar, filtradas))
        return promedio_valido(calibradas)

    def categoria(self) -> str:
        """Devuelve la categoría oficial de calidad del aire para el promedio calibrado."""
        prom = self.promedio_calibrado()
        if prom is None:
            return "Sin datos"
        return categoria_pm25(prom)

    def __str__(self) -> str:
        """Devuelve el texto formateado exactamente con el formato requerido."""
        prom = self.promedio_calibrado()
        prom_redondeado = round(prom, 1) if prom is not None else 0.0
        return f"{self._estacion.codigo}: {prom_redondeado} ug/m3 - {self.categoria()}"

informe = InformeEstacion(estacion_oriente, sensor_a)

print("print(informe):", informe)
print("promedio_calibrado():", informe.promedio_calibrado())
```
